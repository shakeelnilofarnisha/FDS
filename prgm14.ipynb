```python
import numpy as np
import pandas as pd
data = {
    'CustomerID': [1, 2, 3, 4, 5, 6, 7, 8, 9, 9, 10],
    'Age_Group': ['20-25', '30-35', '25-30', '20-25', '35+', '35+', '35+', '20-25', '25-30', '25-30', '30-35'],
    'Rating(1-5)': [4, 5, 6, -1, 3, 3, 4, 7, 2, 2, np.nan],
    'Hotel': ['Ibis', 'LemonTree', 'RedFox', 'LemonTree', 'Ibis', 'Ibys', 'RedFox', 'LemonTree', 'Ibis', 'Ibis', 'RedFox'],
    'FoodPreference': ['veg', 'Non-Veg', 'Veg', 'Veg', 'Vegetarian', 'Non-Veg', 'Vegetarian', 'Veg', 'non-Veg', 'Non-Veg', 'non-Veg'],
    'Bill': [1300, 2000, 1322, 1322, 1234, 989, 1909, 1000, 2999, 3456, 6755],
    'NoOfPax': [2, 3, 2, 2, 2, 2, 2, -1, -10, 3, 4],
    'EstimatedSalary': [40000, 59000, 30000, 30000, 120000, 45000, 122220, 21122, 345673, -99999, 87777]
}
df = pd.DataFrame(data)
print("Original Dataset")
print(df)
print("\nDuplicate Rows")
print(df.duplicated())
print("\nDataset Information")
df.info()
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)
df['CustomerID'] = df['CustomerID'].mask(df['CustomerID'] < 0)
df['Bill'] = df['Bill'].mask(df['Bill'] < 0)
df['EstimatedSalary'] = df['EstimatedSalary'].mask(df['EstimatedSalary'] < 0)
df['NoOfPax'] = df['NoOfPax'].mask((df['NoOfPax'] < 1) | (df['NoOfPax'] > 20))
print("\nUnique Age Groups")
print(df.Age_Group.unique())
print("\nUnique Hotels")
print(df.Hotel.unique())
df['Hotel'] = df['Hotel'].replace('Ibys', 'Ibis')
print("\nUnique Food Preferences")
print(df.FoodPreference.unique())
df['FoodPreference'] = df['FoodPreference'].replace(
    ['Vegetarian', 'veg'], 'Veg'
)
df['FoodPreference'] = df['FoodPreference'].replace(
    'non-Veg', 'Non-Veg'
)
df['EstimatedSalary'] = df['EstimatedSalary'].fillna(
    round(df['EstimatedSalary'].mean())
)
df['NoOfPax'] = df['NoOfPax'].fillna(
    round(df['NoOfPax'].median())
)
df['Rating(1-5)'] = df['Rating(1-5)'].mask(
    (df['Rating(1-5)'] < 1) | (df['Rating(1-5)'] > 5)
)
df['Rating(1-5)'] = df['Rating(1-5)'].fillna(
    round(df['Rating(1-5)'].median())
)
df['Bill'] = df['Bill'].fillna(
    round(df['Bill'].mean())
)
print("\nCleaned Dataset")
print(df)
```
