```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = {
    'Emp_ID': ['E101', 'E102', 'E103', 'E104', 'E105', 'E106',
               'E107', 'E108', 'E109', 'E110', 'E111', 'E112'],

    'Name': ['Arun', 'Divya', 'Karthik', 'Priya', 'Naveen', 'Meena',
             'Rahul', 'Anitha', 'Sanjay', 'Kavya', 'Arun', 'Lakshmi'],

    'Age': [25, 28, 24, np.nan, 31, 27, 29, 26, 30, 25, 25, 28],

    'Gender': ['Male', 'Female', 'M', 'Female', 'Male', 'F',
               'Male', 'Female', 'male', 'Female', 'Male', 'FEMALE'],

    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT', 'HR',
                   'Finance', 'IT', 'HR', 'IT', 'IT', 'finance'],

    'Salary': ['35,000', '42,000', '32,000', '48,000', np.nan,
               '40,000', '450,000', '38,000', '44,000',
               '36,000', '35,000', '43,000'],

    'Experience': [2, 5, 1, 6, 7, 4, 5, np.nan, 6, 2, 2, 4],

    'Performance': [78, 85, 72, 91, 88, 82, 87, 80, 150, 76, 78, 84],

    'City': ['Chennai', 'Madurai', 'Chennai', 'Coimbatore',
             'Chennai', 'Madurai', 'Chennai', 'Coimbatore',
             'Madurai', 'Chennai', 'Chennai', 'Madurai']
}

df = pd.DataFrame(data)

print("--- RAW EMPLOYEE DATABASE ---")
print(df)
```

```python
print("--- 2. DATA PROFILING ---")
print("\nDataset Info:")
df.info()
print(df.isnull().sum())
```

```python
df['Gender'] = df['Gender'].replace({
    'M': 'Male',
    'F': 'Female',
    'male': 'Male',
    'FEMALE': 'Female'
})

# Clean Department
df['Department'] = df['Department'].str.title()

# Convert Salary to numeric
df['Salary'] = df['Salary'].astype(str).str.replace(',', '')

df['Salary'] = pd.to_numeric(
    df['Salary'],
    errors='coerce'
)

print("--- DATA CLEANSING ---")
print(df)
```

```python
lower_limit = 0
upper_limit = 100

# Invalid values converted to NaN
df.loc[
    (df['Performance'] < lower_limit) |
    (df['Performance'] > upper_limit),
    'Performance'
] = np.nan

print("--- CONSTRAINT VALIDATION ---")
print(df)
```

```python

df['Age'] = df['Age'].fillna(df['Age'].median())

df['Salary'] = df['Salary'].fillna(
    df['Salary'].median()
)

df['Experience'] = df['Experience'].fillna(
    df['Experience'].median()
)

df['Performance'] = df['Performance'].fillna(
    df['Performance'].median()
)

print("--- MISSING DATA IMPUTATION ---")
print(df)
```

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 28, 100],
    labels=['Young', 'Middle Aged']
)

# Region Mapping
region_map = {
    'Chennai': 'North TN',
    'Coimbatore': 'West TN',
    'Madurai': 'South TN'
}

df['Region'] = df['City'].map(region_map)

print("--- DATA TRANSFORMATION ---")
print(df[['Emp_ID', 'Name', 'Age', 'City', 'Age_Group', 'Region']])
```

```python
normalizer = MinMaxScaler()
df['Performance_Normalized'] = normalizer.fit_transform(df[['Performance']])

print("--- DATA NORMALIZATION ---")
print(df[[ 'Performance_Normalized']])
```

```python
df_reduced = df.drop(
    columns=['Emp_ID', 'Name']
)

print("--- DATA REDUCTION ---")
print(df_reduced)
```
